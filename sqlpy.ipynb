```python
import pandas as pd
import sqlalchemy as sal

engine = sal.create_engine('mssql://AL15-41\\SQLEXPRESS/naamstesql?driver=ODBC+Driver+17+for+SQL+Server')
conn = engine.connect()
print("conn success")
```

```python
def extract():
    df_orders = pd.read_csv('orders.txt')
    df_returns = pd.read_csv('returns.txt')
    df_returns.rename(columns={'Order Id':'order_id'}, inplace = True)
    return df_orders, df_returns

def transform(df_orders,df_returns):
    df = pd.merge(df_orders,df_returns, left_on = 'order_id', right_on = 'order_id',  how = 'inner')
    return df
def load(df,conn):
    df.to_sql(con = conn, name = 'orders_final', if_exists = 'replace', index = False)
    
   
    
```

```python
df_orders,df_returns = extract()
df = transform(df_orders,df_returns)
load(df,conn)
```

```python
df_sql = pd.read_sql_query('''select city,sum(sales) over(partition by city order by cast(order_date AS DATE)) as tot_sales from orders_final''' , conn)
```
