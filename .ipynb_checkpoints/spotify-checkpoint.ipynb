```python
id = []
track_number = []
name = []
duramtion_s = []
album = []
artist = []
popularity = []

for track in playlist['tracks']['items']:        #its a list
    name.append(songs['track']['name'])
    id.append(songs['track']['id'])
    track_number.append(songs['track']['track_number'])
    duramtion_s.append(songs['track']['duration_ms']/(1000))
    album.append(songs['track']['album']['name'])
    artist.append(songs['track']['artists'][0]['name'])
    popularity.append(songs['track']['popularity'])
```

```python
# Create Dataframes in format {'key1':[list of values], 'key2':[list of 2nd column values]..............}
# this way keys of dict act as column names and values for each column is there inside the 
# list of values

import pandas as pd
df = pd.DataFrame({'name': name, 'id': id, 'track_number': track_number,
                   'duramtion_s': duramtion_s, 'album': album, 'artist': artist, 
                   'popularity': popularity})
```
