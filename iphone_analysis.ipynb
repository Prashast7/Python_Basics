```python
import pandas as pd
df = pd.read_csv('iphone.csv')
df.head()
```

```python
df.columns = df.columns.str.replace(' ','_')
df.head()
```

```python
df[df['Star_Rating'].isna()]
```

```python
mean_value = df['Star_Rating'].mean()
```

```python
df['Star_Rating'].fillna(mean_value, inplace = True)
```

```python
df['Star_Rating'].head()
```

```python
df['Discount_Percentage'] = ((df['Mrp'] - df['Sale_Price']) /df['Mrp']) * 100
```

```python
x = df['Discount_Percentage'].max()
max_discount = df[df['Discount_Percentage'] == x]
max_discount.head()
```

```python
df['storage'] = df['Product_Name'].str.extract(r'(\d+)\s?GB')
```

```python
df.head()
```

```python
top_5 = df.nlargest(5,'Number_Of_Reviews')[['Product_Name','Number_Of_Reviews']]
top_5
```

```python
hl_df = df['Mrp'].max()-df['Mrp'].min()
print(hl_df)
```

```python
third_df = df.nlargest(3,'Mrp')[['Product_Name','Mrp']].iloc[-1]
third_df
```

```python
avg_df = df[df['Mrp']>100000]['Mrp'].mean()
avg_df
```

```python
df_128 = df[df['Product_Name'].str.contains('128 GB', case= False, na = False)]
df_128['Rating_To_Review_Ratio'] = (df_128['Number_Of_Ratings'] /	df_128['Number_Of_Reviews'])
max_ratio_product = df_128.sort_values('Rating_To_Review_Ratio', ascending=False).iloc[0]['Product_Name']

max_ratio_product
```
