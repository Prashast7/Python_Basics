```python
import pandas as pd
```

```python
df = pd.read_csv('returns.txt')
```

```python
df.head(5)
```

```python
dfo = pd.read_csv('orders.txt')
```

```python

df.rename(columns={'Return Reason':'return_reason'}, inplace = True)
x = df['return_reason'].unique()
print(x)
```

```python
# write a program to get category wise sales of orders that were not returned
return_df = df['Order Id'].unique()
orders_df = dfo[~dfo['order_id'].isin(return_df)]
orders_df.groupby('category').agg({'sales':'sum'})
```

```python
df.rename(columns={'Order Id':'order_id'}, inplace = True)
```

```python
#write a program to get city wise count of return orders
city_df = pd.merge(dfo, df, how='left', on = 'order_id')
city_df['city'].value_counts()
```

```python
# write a program to print cities where we have all 3 kinds of returns (others,bad quality,wrong items)
city_return = pd.merge(dfo,df,how='left',on = 'order_id')
city_ret_counts = city_return.groupby('city')['return_reason'].nunique()
city_ret_counts[city_ret_counts.values > 1]
```

```python
# write a program to find top 3 cities by sales
dfo.groupby('city')['sales'].sum().sort_values(ascending=False).head(3)
```
